```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("notas_estudo.csv")
print(df)
```

```python
sns.scatterplot(x="horas_estudo_semana", y="nota_prova", data=df)
plt.title("Horas de estudo x Nota da prova")
plt.xlabel("Horas de estudo por semana")
plt.ylabel("Nota da prova")
plt.show()
```
