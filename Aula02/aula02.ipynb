```python
import numpy as np
import time

n = 10_000_000
lista_python = list(range(n))
array_numpy = np.arange(n)

inicio = time.time()
resultado_python = [x * 2 for x in lista_python]
fim = time.time()
print(f"Python puro: {fim - inicio:.4f} segundos")

inicio = time.time()
resultado_numpy = array_numpy * 2
fim = time.time()
print(f"NumPy: {fim - inicio:.4f} segundos")
```
