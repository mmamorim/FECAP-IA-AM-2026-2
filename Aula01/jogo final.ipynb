## Bloco 1 — Como representar um animal?

```python
cachorro = {
    "nome": "Cachorro",
    "voa": False,
    "nada": True,
    "tem_pelos": True,
    "e_domestico": True,
    "bota_ovos": False,
}

cachorro
```

## Bloco 2 — Como representar vários animais?

```python
animais = [
    {"nome": "Cachorro", "voa": False, "nada": True,  "tem_pelos": True,  "e_domestico": True,  "bota_ovos": False},
    {"nome": "Aguia",    "voa": True,  "nada": False, "tem_pelos": False, "e_domestico": False, "bota_ovos": True},
    {"nome": "Pato",     "voa": True,  "nada": True,  "tem_pelos": False, "e_domestico": True,  "bota_ovos": True},
    {"nome": "Gato",     "voa": False, "nada": False, "tem_pelos": True,  "e_domestico": True,  "bota_ovos": False},
]

len(animais)
```

## Bloco 3 — Como percorrer essa coleção?

```python
for animal in animais:
    print(animal["nome"])
```

## Bloco 4 — Como eliminar quem não atende uma característica?

```python
candidatos = []
for animal in animais:
    if animal["voa"] == True:
        candidatos.append(animal)

print(f"Restam {len(candidatos)} candidatos")
for c in candidatos:
    print(c["nome"])
```

## Bloco 5 — Como combinar dois critérios ao mesmo tempo?

```python
candidatos = []
for animal in animais:
    if animal["voa"] == True and animal["e_domestico"] == True:
        candidatos.append(animal)

print(f"Restam {len(candidatos)} candidatos")
for c in candidatos:
    print(c["nome"])
```

## Bloco 6 — Como transformar isso em uma função reutilizável?

```python
def filtrar(lista, atributo, valor):
    resultado = []
    for item in lista:
        if item[atributo] == valor:
            resultado.append(item)
    return resultado

candidatos = filtrar(animais, "nada", True)
print(f"Restam {len(candidatos)} candidatos")
for c in candidatos:
    print(c["nome"])
```

## Bloco 7 — Como perguntar algo para quem está usando o programa?

Até agora, todas as respostas (`True`/`False`) foram **escritas por nós, no código**. Mas e se quiséssemos que a própria pessoa jogando respondesse?

Python tem uma função pronta pra isso: `input()`. Ela pausa o programa, mostra uma pergunta na tela, e espera alguém digitar uma resposta.

**Atenção:** `input()` sempre devolve uma **string** (texto) — mesmo que a pessoa digite "sim" ou um número.

```python
nome_pessoa = input("Qual é o seu nome? ")
print(f"Oi, {nome_pessoa}! Vamos jogar?")
```

## Bloco 8 — Usando o `input()` para filtrar de verdade

Agora vamos conectar o `input()` com a função `filtrar()` que já criamos. A resposta digitada (texto) precisa virar `True`/`False` antes de comparar — por isso o `.strip().lower() == "sim"` no final: ele limpa espaços em branco, deixa tudo minúsculo, e compara com `"sim"`, virando um booleano.

```python
resposta_texto = input("O animal voa? (sim/nao): ")
resposta = resposta_texto.strip().lower() == "sim"

candidatos = filtrar(animais, "voa", resposta)
print(f"Restam {len(candidatos)} candidatos: {[c['nome'] for c in candidatos]}")
```

## Bloco 9 — Como carregar animais de um arquivo externo?

Digitar animal por animal, na mão, funciona pra 4 exemplos — mas não escala. Bancos de dados de verdade ficam em **arquivos externos**, que o programa carrega quando precisa.

Vamos usar um arquivo `animais_completo.json`, com dezenas de animais já prontos, no mesmo formato (dicionários com os mesmos atributos) que vimos até agora.

**Passo 1 — subir o arquivo pro Colab.** Rode a célula abaixo e selecione o arquivo `animais_completo.json` no seu computador quando a janela abrir.

```python
from google.colab import files

enviado = files.upload()  # abre uma janela para você escolher o arquivo animais_completo.json
```

**Passo 2 — ler o conteúdo do arquivo.** O módulo `json` (parte do Python, não precisa instalar nada) sabe transformar o texto do arquivo direto em lista de dicionários — a mesma estrutura que já estávamos usando.

```python
import json

with open("animais_completo.json", "r", encoding="utf-8") as arquivo:
    animais = json.load(arquivo)

print(f"Carregamos {len(animais)} animais!")
animais[:3]  # espiando os 3 primeiros
```

## Bloco 10 — Juntando tudo: o jogo completo

Agora temos todas as peças: a lista grande de animais (carregada do arquivo), a função `filtrar()`, e o `input()` pra perguntar. Vamos juntar tudo num loop que pergunta, filtra, pergunta de novo — até sobrar só um candidato (ou os dois casos especiais: ninguém bate, ou ainda há empate).

```python
perguntas = {
    "voa": "O animal voa?",
    "nada": "O animal nada?",
    "tem_pelos": "O animal tem pelos?",
    "e_domestico": "O animal e domestico?",
    "bota_ovos": "O animal bota ovos?",
    "tem_quatro_patas": "O animal tem quatro patas?",
    "vive_na_agua": "O animal vive na agua?",
    "e_carnivoro": "O animal e carnivoro?",
    "tem_cauda": "O animal tem cauda?",
}

candidatos = animais.copy()

for atributo, pergunta in perguntas.items():
    if len(candidatos) <= 1:
        break
    resposta_texto = input(f"{pergunta} (sim/nao): ")
    resposta = resposta_texto.strip().lower() == "sim"
    candidatos = filtrar(candidatos, atributo, resposta)
    print(f"Restam {len(candidatos)} candidato(s)")

print()
if len(candidatos) == 1:
    print(f"O animal que voce pensou e: {candidatos[0]['nome']}! 🎉")
elif len(candidatos) == 0:
    print("Nao encontrei nenhum animal com essas caracteristicas. 🤔")
else:
    nomes = [c["nome"] for c in candidatos]
    print(f"Ainda restam algumas possibilidades: {', '.join(nomes)}")
```
