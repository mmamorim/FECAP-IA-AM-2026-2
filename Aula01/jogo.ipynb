## Bloco 1 — Como representar um animal?

```python
cachorro = {
    "nome": "Cachorro",
    "voa": False,
    "nada": True,
    "tem_pelos": True,
    "e_domestico": True,
    "bota_ovos": False,
}

print(cachorro)
```

```python
animais = [
    {"nome": "Cachorro", "voa": False, "nada": True,  "tem_pelos": True,  "e_domestico": True,  "bota_ovos": False},
    {"nome": "Aguia",    "voa": True,  "nada": False, "tem_pelos": False, "e_domestico": False, "bota_ovos": True},
    {"nome": "Pato",     "voa": True,  "nada": True,  "tem_pelos": False, "e_domestico": True,  "bota_ovos": True},
    {"nome": "Gato",     "voa": False, "nada": False, "tem_pelos": True,  "e_domestico": True,  "bota_ovos": False},
]

len(animais)
```

```python
for animal in animais:
    print(animal["nome"])
```

```python
candidatos = []
for animal in animais:
    if animal["voa"] == True:
        candidatos.append(animal)

print(f"Restam {len(candidatos)} candidatos")
for c in candidatos:
    print(c["nome"])
```

```python
candidatos = []
for animal in animais:
    if animal["voa"] == True and animal["e_domestico"] == True:
        candidatos.append(animal)

print(f"Restam {len(candidatos)} candidatos")
for c in candidatos:
    print(c["nome"])
```

```python
def filtrar(lista, atributo, valor):
    resultado = []
    for item in lista:
        if item[atributo] == valor:
            resultado.append(item)
    return resultado

candidatos = filtrar(animais, "nada", True)
print(f"Restam {len(candidatos)} candidatos")
for c in candidatos:
    print(c["nome"])
```

```python
resposta_texto = input("O animal voa? (sim/nao): ")
resposta = resposta_texto.strip().lower() == "sim"

candidatos = filtrar(animais, "voa", resposta)
print(f"Restam {len(candidatos)} candidatos: {[c['nome'] for c in candidatos]}")
```

```python
import json

with open("animais_completo.json", "r", encoding="utf-8") as arquivo:
    animais = json.load(arquivo)

print(f"Carregamos {len(animais)} animais!")
animais[:3]  # espiando os 3 primeiros
```

```python
perguntas = {
    "voa": "O animal voa?",
    "nada": "O animal nada?",
    "tem_pelos": "O animal tem pelos?",
    "e_domestico": "O animal e domestico?",
    "bota_ovos": "O animal bota ovos?",
    "tem_quatro_patas": "O animal tem quatro patas?",
    "vive_na_agua": "O animal vive na agua?",
    "e_carnivoro": "O animal e carnivoro?",
    "tem_cauda": "O animal tem cauda?",
}

candidatos = animais.copy()

for atributo, pergunta in perguntas.items():
    if len(candidatos) <= 1:
        break
    resposta_texto = input(f"{pergunta} (sim/nao): ")
    resposta = resposta_texto.strip().lower() == "sim"
    candidatos = filtrar(candidatos, atributo, resposta)
    print(f"Restam {len(candidatos)} candidato(s)")

print()
if len(candidatos) == 1:
    print(f"O animal que voce pensou e: {candidatos[0]['nome']}! 🎉")
elif len(candidatos) == 0:
    print("Nao encontrei nenhum animal com essas caracteristicas. 🤔")
else:
    nomes = [c["nome"] for c in candidatos]
    print(f"Ainda restam algumas possibilidades: {', '.join(nomes)}")
```
